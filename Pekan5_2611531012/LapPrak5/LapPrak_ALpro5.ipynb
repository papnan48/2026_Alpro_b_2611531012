## 2.1 Dasar Teori
1. **Perulangan `for`:** Perulangan ini digunakan untuk mengiterasi suatu barisan (*sequence* seperti *list, tuple, string*) atau dieksekusi sejumlah iterasi numerik yang dipatok menggunakan fungsi tertentu. Berbeda dengan *while-loop* yang bergantung murni pada kondisi Boolean *(True/False)*, *for-loop* memiliki ujung iterasi yang jauh lebih terprediksi.
2. **Fungsi `range()`:** Sering mendampingi `for`, fungsi ini menghasilkan deret bilangan berurutan. Format utamanya adalah `range(start, stop, step)`. Nilai *stop* bersifat eksklusif (tidak diikutsertakan dalam eksekusi) sehingga untuk mencapai iterasi `N`, argumen yang diberikan idealnya adalah `N+1`.
3. **Nested Loop (Perulangan Bersarang):** Merupakan skenario peletakan blok perulangan di dalam kurung perulangan yang lain. Variabel perulangan di bagian luar (*outer loop*) umumnya mengendalikan pergantian dan lompatan baris (vertikal), sedangkan perulangan di sisi dalam (*inner loop*) mengendalikan perulangan karakter per karakter pada struktur baris tersebut secara memanjang ke samping (horizontal).

## 2.2 Kode Program

### 2.2.1 Program "perulangan_for1_2611531012.py"


```python
# Buat file dengan nama perulangan_for1_2611531012.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1012
# program ini menggunakan fungsi input()

ulang_1012 = int(input("Masukkan jumlah perulangan: "))

for i_1012 in range(ulang_1012):
    print(f"Perulangan ke-{i_1012}")
```

**Penjelasan :**
Program ini menunjukkan konsep paling dasar dari perulangan `for`. Dengan menggunakan fungsi `range(ulang)`, program melakukan iterasi mulai dari indeks `0` hingga batas `ulang - 1`. Pada setiap iterasinya, program mencetak string "Perulangan ke-" yang digabungkan dengan nilai indeks iterasi saat itu.

### 2.2.2 Program "perulangan_for2_2611531012.py"


```python
# Buat file dengan nama perulangan_for2_2611531012.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1012
# program ini menggunakan fungsi input()

ulang_1012 = int(input("Masukkan jumlah perulangan: "))
print("Perulangan ke-0 sampai ke-", ulang_1012-1)
for i_1012 in range(ulang_1012):
    print(i_1012, end=" ")
print()
print("Perulangan ke-1 sampai ke-", ulang_1012)
for i_1012 in range(1, ulang_1012+1):
    print(i_1012, end=" ")
```

**Penjelasan:**

Program ini mendemonstrasikan cara memanipulasi parameter batas awal dan batas akhir pada fungsi `range()`. Perulangan pertama menggunakan `range(ulang)` yang secara bawaan dimulai dari angka 0. Perulangan kedua memodifikasi rentang menjadi `range(1, ulang + 1)` agar iterasi dimulai dari angka 1 dan berakhir tepat pada nilai input pengguna. Parameter `end=" "` ditambahkan agar hasil cetakan tersusun menyamping (horizontal).

### 2.2.3 Program "perulangan_for3_2611531012.py"


```python
# Buat file dengan nama perulangan_for3_2611531012.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1012
# program ini menggunakan fungsi input()

ulang_1012 = int(input("Masukkan jumlah perulangan: "))

jumlah_1012 = 0
for i_1012 in range(1, ulang_1012 + 1):
    print(i_1012, end=" ")
    jumlah_1012 = jumlah_1012 + i_1012

    if i_1012 < ulang_1012:
        print(" + ", end=" ")
    else:
        print(" = ", jumlah_1012, end=" ")

print()
print("Jumlah =", jumlah_1012)
```

**Penjelasan:**

Program ini mengimplementasikan logika akumulasi dengan bantuan variabel penyimpan `jumlah`. Pada setiap perulangan, angka yang sedang diiterasi akan dicetak dan ditambahkan ke dalam variabel `jumlah`. Selain itu, program memanfaatkan percabangan `if-else` untuk menentukan kapan harus mencetak simbol `+` (jika iterasi masih berlangsung) dan kapan harus mencetak simbol `=` beserta hasil totalnya (jika iterasi sudah mencapai angka terakhir).

### 2.2.4 Program "jumlah_genap_2611531012.py"


```python
# Buat file dengan nama jumlah_genap_2611531012.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1012
# program ini menggunakan fungsi input()

ulang_1012 = int(input("Masukkan nilai batas: "))

jumlah_1012 = 0
for i_1012 in range(1, ulang_1012 + 1):
    if i_1012 % 2 == 0:
        print(i_1012, end=" ")
        jumlah_1012 = jumlah_1012 + i_1012

        if i_1012 < ulang_1012:
            print(" + ", end=" ")
        else:
            print(" = ", jumlah_1012, end=" ")
print()
print("Jumlah =", jumlah_1012)
```

**Penjelasan :**
Program ini merupakan pengembangan dari logika akumulasi sebelumnya dengan penambahan filter kondisional. Terdapat struktur `if i % 2 == 0:` di dalam perulangan yang berfungsi untuk mengecek sisa bagi. Dengan kondisi ini, program hanya akan mencetak, memproses, dan menjumlahkan angka-angka yang bernilai genap saja, sedangkan angka ganjil akan dilewati (*skip*).

### 2.2.5 Program "nested_for1_2611531012.py"


```python
# Buat file dengan nama nested_for1_2611531012.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1012
# program ini menggunakan fungsi input()

batas_1012 = int(input("Masukkan nilai batas: "))
for line_1012 in range(1, batas_1012 + 1):
    for j_1012 in range(1, (-1 * line_1012 + batas_1012) + 1):
        print(".", end=" ")
    print(line_1012)
```

**Penjelasan :**
Program ini memperkenalkan perulangan bersarang (*nested loop*). *Outer loop* (perulangan luar) bertugas mengatur perpindahan baris, sedangkan *inner loop* (perulangan dalam) bertugas mencetak karakter titik (`.`). Jumlah karakter titik yang dicetak bersifat dinamis dan semakin menurun pada setiap barisnya, dikendalikan oleh formula matematika `(-1 * line + batas)`. Pada bagian ujung tiap baris, nilai baris tersebut dicetak.

### 2.2.6 Program "nested_for2_2611531012.py"


```python
# Buat file dengan nama nested_for2_2611531012.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1012
# Program ini menggunakan fungsi input()

batas_1012 = int(input("Masukkan nilai batas: "))
for i_1012 in range(1, batas_1012 + 1):
    for j_1012 in range(1, batas_1012 + 1):
        print("*", end=" ")
    print() # pindah ke baris berikutnya
```

**Penjelasan :**
Program ini membentuk pola bangun datar persegi menggunakan karakter asterisk (`*`). *Outer loop* melakukan iterasi untuk menentukan jumlah baris ke bawah, sementara *inner loop* melakukan iterasi untuk mencetak karakter bintang pada setiap kolom menyamping. Sebuah perintah `print()` kosong diletakkan di akhir *outer loop* sebagai pemicu (*trigger*) untuk pindah ke baris baru.

### 2.2.7 Program "nested_for3_2611531012.py"


```python
# Buat file dengan nama nested_for3_2611531012.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1012
# Program ini menggunakan fungsi input()

batas_1012 = int(input("Masukkan nilai batas: "))
for i_1012 in range(batas_1012 + 1):
    for j_1012 in range(batas_1012 + 1):
        print(i_1012 + j_1012, end=" ")
    print() # pindah ke baris berikutnya
```

**Penjelasan :**
Program ini menampilkan matriks numerik dua dimensi yang nilainya terus meningkat. Logikanya berpusat pada relasi antara indeks baris (`i`) dan indeks kolom (`j`). Di dalam *inner loop*, program mencetak hasil penjumlahan dari posisi baris dan posisi kolom (`i + j`). Hal ini menghasilkan deret angka yang membentuk pola diagonal yang berurutan.

### 2.2.8 Program "nested_for4_2611531012.py"


```python
# Buat file dengan nama nested_for4_2611531012.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1012
# Program ini menggunakan fungsi input()

tinggi_1012 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_1012 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_1012 = tinggi_1012
    c_1012 = a_1012
    lebar_1012 = (2 * tinggi_1012) - 2

    for i_1012 in range(1, tinggi_1012 + 1):
        b_1012 = c_1012 + 1

        for j_1012 in range(1, lebar_1012 + 1):

            # Baris atas dan bawah
            if i_1012 == 1 or i_1012 == tinggi_1012:
                if j_1012 == 1 or j_1012 == lebar_1012:
                    print("#", end="")
                else:
                    print("=", end="")
                    
            # Baris isi
            else:
                if j_1012 == 1 or j_1012 == lebar_1012:
                    print("|", end="")
                else:
                    if j_1012 == c_1012:
                        print("<", end="")
                    elif j_1012 == b_1012:
                        print(">", end="")
                    elif j_1012 == (lebar_1012 - c_1012):
                        print("<", end="")
                    elif j_1012 == (lebar_1012 - c_1012 + 1):
                        print(">", end="")
                    elif j_1012 > b_1012 and j_1012 < (lebar_1012 - c_1012):
                        print(".", end="")
                    else:
                        print(" ", end="")
        print()

        # Logika asli Java
        a_1012 -= 2

        if a_1012 <= 0:
            c_1012 = (-a_1012) + 2
        else:
            c_1012 = a_1012
```

**Penjelasan :**
Program ini menggunakan *nested loop* kompleks untuk membentuk sebuah bangun simetris dua dimensi yang dilengkapi dengan bingkai luar. *Outer loop* menangani tinggi pola, sementara *inner loop* menyusun setiap piksel karakter secara horizontal. Program ini sangat bergantung pada struktur percabangan tingkat lanjut (`if-elif-else`) dan variabel kalkulasi geometri (`a`, `b`, `c`, `lebar`) untuk menentukan letak koordinat yang pas guna mencetak batas pagar (`#`, `=`), dinding (`|`), maupun corak bagian dalam (`<`, `>`, `.`, atau spasi kosong).

### 2.2.9 Program "latihan5_2611531012.py"


```python
# Soal: Buatlah proghgram untuk menampilkan output dibawah menggunakan perulangan for
# Ex: Masukkan tinggi segitiga: 5
# Output:
#     *
#    * *
#   * * *
#  * * * *
# * * * * *

tinggi_1012 = int(input("Masukkan tinggi segitiga: "))

for i_1012 in range(1, tinggi_1012 + 1):
    for j_1012 in range(tinggi_1012 - i_1012):
        print(" ", end="")
    for k_1012 in range(i_1012):
        print("* ", end="")
    print()
```

**Penjelasan :**
Program ini menggunakan perulangan bersarang untuk mencetak pola segitiga sama kaki (piramida) menggunakan karakter bintang (`*`). *Outer loop* bertugas mengendalikan jumlah baris dari atas ke bawah sesuai dengan tinggi yang diinputkan pengguna. Di dalam *outer loop* tersebut, terdapat dua *inner loop* yang berjalan secara berurutan: *inner loop* pertama mencetak spasi kosong yang jumlahnya semakin menurun di tiap baris agar pola terdorong ke tengah, sedangkan *inner loop* kedua mencetak karakter bintang yang diikuti satu spasi (`* `) dengan jumlah yang semakin meningkat. Perpaduan spasi penyeimbang dan penambahan karakter bintang ini menghasilkan bentuk bangun piramida yang simetris dan rapi.

## Tugas Praktikum

**Soal :**

Studi Kasus: Pola Jam Pasir Kristal Palindromik Berbingkai (Framed Palindromic Crystal Hourglass)
Mahasiswa diminta merancang program konsol interaktif untuk mencetak pola bangun geometri dua dimensi berupa Jam Pasir Kristal Palindromik yang terbungkus rapi di dalam bingkai pembatas berpagar. Berbeda dengan pola dasar, program ini menuntut pembentukan deret angka yang bergerak dinamis menurun dan menaik secara palindromik pada setiap baris serta diselaraskan dengan poros kristal di pusat bangun.
Program menerima masukan nilai integer positif N dari pengguna yang merepresentasikan ukuran skala jam pasir (skala angka tertinggi pada puncak dan alas).

**Kode Jawaban :**

a. Kode Program "tugas5_2611531012.py"

```python
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK ===")
n_1012 = int(input("Masukkan ukuran skala jam pasir (N): "))

# 1. Bingkai Pembatas Horizontal (Border Atas)
print("#", end="")
for i_1012 in range(4 * n_1012 + 5):
    print("=", end="")
print("#", end="")
print() # Pindah baris

# 2. Fase 1: Jam Pasir Atas (Reduksi Angka Menurun: Baris N turun s.d. 1)
for baris_1012 in range(n_1012, 0, -1):
    print("| ", end="") 
    
    # Spasi penyeimbang kiri
    for spasi_1012 in range(2 * (n_1012 - baris_1012)):
        print(" ", end="")
        
    # Deret angka mundur
    for angka_1012 in range(baris_1012, 0, -1):
        print(angka_1012, end=" ")
        
    # Poros kristal tengah
    print("<*>", end="")
    
    # Deret angka maju
    for angka_1012 in range(1, baris_1012 + 1):
        print(" ", end="")
        print(angka_1012, end="")
        
    # Spasi penyeimbang kanan
    for spasi_1012 in range(2 * (n_1012 - baris_1012)):
        print(" ", end="")
        
    print(" |", end="")
    print() # Pindah baris

# 3. Fase 2: Poros Titik Pusat Jam Pasir (Singularity)
print("|", end="")
for spasi_1012 in range(2 * n_1012 + 1):
    print(" ", end="")
print("<*>", end="")
for spasi_1012 in range(2 * n_1012 + 1):
    print(" ", end="")
print("|", end="")
print() # Pindah baris

# 4. Fase 3: Jam Pasir Bawah (Ekspansi Angka Menaik: Baris 1 naik s.d. N)
for baris_1012 in range(1, n_1012 + 1):
    print("| ", end="")
    
    # Spasi penyeimbang kiri
    for spasi_1012 in range(2 * (n_1012 - baris_1012)):
        print(" ", end="")
        
    # Deret angka mundur
    for angka_1012 in range(baris_1012, 0, -1):
        print(angka_1012, end=" ")
        
    # Poros kristal tengah
    print("<*>", end="")
    
    # Deret angka maju
    for angka_1012 in range(1, baris_1012 + 1):
        print(" ", end="")
        print(angka_1012, end="")
        
    # Spasi penyeimbang kanan
    for spasi_1012 in range(2 * (n_1012 - baris_1012)):
        print(" ", end="")
        
    print(" |", end="")
    print() # Pindah baris

# 5. Bingkai Pembatas Horizontal (Border Bawah)
print("#", end="")
for i_1012 in range(4 * n_1012 + 5):
    print("=", end="")
print("#", end="")
print() # Pindah baris
```

**Penjelasan :**

Program ini merupakan implementasi *nested loop* (perulangan bersarang) tingkat lanjut untuk mencetak bangun geometri dua dimensi secara dinamis tanpa menggunakan trik manipulasi string (seperti perkalian karakter). Logika program ini dibagi menjadi beberapa fase eksekusi:

*   **Bingkai Pembatas Horizontal:** Menggunakan perulangan tunggal untuk mencetak karakter garis ganda (`=`) sebanyak `4 * N + 5` yang diapit oleh karakter pagar (`#`) di ujung kiri dan kanannya. Ini dieksekusi di awal untuk bingkai atas dan di akhir program untuk bingkai bawah.
*   **Fase 1 (Jam Pasir Atas):** Menggunakan *outer loop* yang berjalan mundur (dekremen) dari `N` hingga 1. Di dalam *outer loop* ini, terdapat serangkaian *inner loop* yang mencetak karakter secara horizontal dengan urutan: dinding kiri (`|`), spasi penyeimbang kiri menggunakan formula matematis `2 * (N - baris)`, deret angka menurun (dari `baris` ke 1), karakter poros tengah (`<*>`), deret angka menaik (dari 1 ke `baris`), spasi penyeimbang kanan, dan dinding kanan (`|`). 
*   **Fase 2 (Poros Titik Pusat):** Mencetak titik singularity tepat di tengah bangun. Bagian ini tidak menggunakan *outer loop* baris, melainkan langsung menggunakan perulangan untuk mencetak spasi penyeimbang sebanyak `2 * N + 1` di sisi kiri dan kanan poros kristal (`<*>`).
*   **Fase 3 (Jam Pasir Bawah):** Merupakan cerminan dari Fase 1. *Outer loop* berjalan maju (inkremen) dari 1 hingga `N`. Serangkaian *inner loop* di dalamnya mengulangi logika pencetakan spasi dan angka yang sama persis dengan Fase 1, namun karena iterasi barisnya bertambah, pola yang dihasilkan akan semakin berekspansi (melebar) ke bawah.
*   **Pindah Baris:** Setiap perulangan *inner loop* yang mencetak isi bangun selalu ditahan dalam baris yang sama menggunakan parameter `end=""`. Pemindahan baris baru hanya dilakukan dengan memanggil `print()` kosong persis di ujung setiap *outer loop*.
