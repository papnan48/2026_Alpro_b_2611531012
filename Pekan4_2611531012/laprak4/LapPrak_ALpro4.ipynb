## 2.1 Dasar Teori
Struktur kontrol percabangan di Python digunakan untuk menentukan blok kode mana yang akan dieksekusi berdasarkan evaluasi ekspresi *boolean* (Benar/Salah). Terdapat beberapa bentuk arsitektur percabangan dasar:

1. **If Tunggal**: Struktur paling sederhana yang hanya akan mengeksekusi blok kode di bawahnya jika kondisi bernilai `True`.
2. **If-Else**: Memberikan alternatif jalan (blok kode lain) apabila kondisi utama bernilai `False`.
3. **If-Elif-Else**: Digunakan untuk mengevaluasi lebih dari dua kemungkinan. Kondisi dievaluasi secara berurutan (sekuensial) dari atas ke bawah.
4. **Multi-If Terpisah**: Rangkaian statement `if` yang ditulis secara independen (tidak saling terkait dengan `elif`). Cocok untuk skenario di mana banyak kondisi dapat terpenuhi secara bersamaan (misalnya akumulasi diskon).
5. **Match-Case**: Fitur *Structural Pattern Matching* yang hadir mulai dari Python 3.10. Sangat efektif menggantikan rangkaian `if-elif` panjang saat mencocokkan satu nilai variabel terhadap sejumlah konstanta pasti.
6. **Operator Logika**: Operator seperti `and`, `or`, dan `not` yang digabungkan ke dalam percabangan untuk membentuk kondisi *majemuk*.

### 2.2.3 Program "multi_if1_2611531012.py"

### 2.2.4 Program "if_elif_else1_2611531012.py"

### 2.2.5 Program "multi_if2_2611531012.py"


```python
# Buat file dengan nama multi_if2_2611531012.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit NIM terakhir contoh: total_belanja_1012
# program ini menggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja_1012 = float(input("Input Total Belanja Anda (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_1012 = input("Apakah Anda member? (y/t): ").strip().lower()
is_member_1012 = input_member_1012 in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_1012 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_1012 = input_promo_1012 in ["y", "ya"]

total_diskon_persen_1012 = 0

# Multi-IF terpisah: setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_1012 > 1000000:
    total_diskon_persen_1012 += 10 # Diskon belanja besar

if is_member_1012:
    total_diskon_persen_1012 += 5 # Diskon member

if kode_promo_valid_1012:
    total_diskon_persen_1012 += 15 # Diskon voucher

# menghitung nominal diskon dan total bayar
nominal_diskon_1012 = total_belanja_1012 * (total_diskon_persen_1012 / 100)
total_bayar_1012 = total_belanja_1012 - nominal_diskon_1012

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon : {total_diskon_persen_1012}% (Rp {nominal_diskon_1012:,.0f})")
print(f"Total Bayar  : Rp {total_bayar_1012:,.0f}")

print(f"Total diskon yang anda dapatkan: {total_diskon_persen_1012}%")
# Output: Total diskon yang anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

### 2.2.7 Program "match_case1_2611531012.py"


```python
# Buat file dengan nama match_case1_2611531012.py
# Buat program untuk match case
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1012
# Program ini menggunakan fungsi input()
# program konversi angka menjadi nama bulan

bulan_1012 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_1012:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

Pengguna memasukkan variabel bertipe *integer*, lalu alur perbandingan lompat ke nilai `case` yang identik. Penggunaan blok `case _` berfungsi layaknya `else`, di mana ia menangani seluruh probabilitas input di luar opsi 1-12.

## Tugas Praktikum

**Soal :**

Studi Kasus: Sistem Loket Terpadu & Audit Transaksi Ekspedisi Wahana
Sebuah wahana rekreasi petualangan membutuhkan sistem loket berbasis konsol untuk menentukan jenis wahana, memverifikasi kelaikan pengunjung dengan kondisi majemuk (logical operators), menghitung diskon akumulatif menggunakan multi-if terpisah, serta mencetak audit status operasional.
Catatan Materi: Seluruh alur program disusun murni menggunakan 4 pilar percabangan yang diajarkan pada modul Pekan 4: (1) if tunggal, (2) if-else, (3) if-elif-else dengan operator logika (and, or, !=), (4) Multi-if terpisah untuk diskon akumulasi, dan (5) match-case (1 s.d. 5 beserta case _). Tidak ada nested-if (if di dalam if).

**Kode Jawaban :**

a. Kode Program "Tugas4_2611531012.py"

```python
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")
nama_1012 = input("Masukkan Nama Pengunjung        : ")
umur_1012 = int(input("Input umur anda                 : "))

# Mengambil huruf pertama dari input SIM untuk validasi
sim_input_1012 = input("Apakah Anda Sudah Punya SIM C (y/t): ").strip().lower()
sim_1012 = sim_input_1012[0] if len(sim_input_1012) > 0 else 't'

print("\nPilihan Paket Wahana (1-5):")
print("  1. Safari Rimba         (Rp 50,000)")
print("  2. Arung Jeram          (Rp 75,000)")
print("  3. Motor ATV Ekstrim    (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP       (Rp 220,000)")

paket_1012 = int(input("Masukkan nomor paket (1-5)      : "))
jumlah_tiket_1012 = int(input("Masukkan jumlah tiket           : "))
member_1012 = input("Apakah Anda member? (y/t)       : ").strip().lower()
promo_1012 = input("Apakah kode promo valid? (y/t)  : ").strip().lower()

# Variabel flag untuk menghentikan eksekusi tanpa exit() atau nested-if
validasi_1012 = True

# If Tunggal untuk validasi kelogisan tiket
if jumlah_tiket_1012 <= 0:
    print("Peringatan: Kuota tiket tidak valid!")
    validasi_1012 = False

harga_satuan_1012 = 0
nama_wahana_1012 = ""

# Match - Case beserta Case _
match paket_1012:
    case 1:
        nama_wahana_1012 = "Safari Rimba"
        harga_satuan_1012 = 50000
    case 2:
        nama_wahana_1012 = "Arung Jeram"
        harga_satuan_1012 = 75000
    case 3:
        nama_wahana_1012 = "Motor ATV Ekstrim"
        harga_satuan_1012 = 120000
    case 4:
        nama_wahana_1012 = "Roller Coaster Kilat"
        harga_satuan_1012 = 100000
    case 5:
        nama_wahana_1012 = "All-Access VIP"
        harga_satuan_1012 = 220000
    case _:
        print("Paket wahana tidak valid!")
        validasi_1012 = False

# Jika validasi_1012 bernilai False (tiket <= 0 atau paket tidak valid), 
# maka program akan melewati seluruh evaluasi di bawah ini secara otomatis.

if validasi_1012:
    print("\n--- KELAYAKAN PENGENDARA WAHANA ---")

# If - Elif - Else terstruktur tanpa Nested If (if di dalam if)
if validasi_1012 and paket_1012 == 3 and umur_1012 >= 17 and sim_1012 == 'y':
    print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
elif validasi_1012 and paket_1012 == 3 and umur_1012 >= 17 and sim_1012 != 'y':
    print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
elif validasi_1012 and paket_1012 == 3 and umur_1012 < 17 and sim_1012 == 'y':
    print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
elif validasi_1012 and paket_1012 == 3:
    print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")
elif validasi_1012 and umur_1012 >= 10:
    print(f"Status Akses: Umur Anda mencukupi untuk wahana {nama_wahana_1012}.")
elif validasi_1012:
    print(f"Status Akses: Anda belum cukup umur untuk wahana {nama_wahana_1012} (wajib didampingi orang tua).")

# Multi-IF terpisah untuk diskon akumulatif
subtotal_1012 = harga_satuan_1012 * jumlah_tiket_1012
total_diskon_persen_1012 = 0

if validasi_1012 and subtotal_1012 >= 200000:
    total_diskon_persen_1012 += 10
if validasi_1012 and member_1012 in ['y', 'ya']:
    total_diskon_persen_1012 += 5
if validasi_1012 and promo_1012 in ['y', 'ya']:
    total_diskon_persen_1012 += 15
if validasi_1012 and jumlah_tiket_1012 >= 5:
    total_diskon_persen_1012 += 5

# Kalkulasi Nominal Diskon dan Evaluasi Akhir
if validasi_1012:
    nominal_diskon_1012 = subtotal_1012 * (total_diskon_persen_1012 / 100)
    total_bayar_1012 = subtotal_1012 - nominal_diskon_1012

    print("\n--- Rincian Pembayaran ---")
    print(f"Subtotal Belanja : Rp {subtotal_1012:,.0f}")
    print(f"Total Diskon     : {total_diskon_persen_1012}% (Rp {nominal_diskon_1012:,.0f})")
    print(f"Total Bayar      : Rp {total_bayar_1012:,.0f}")

# Evaluasi if - else penutup audit
if validasi_1012 and total_bayar_1012 > 300000:
    print("Catatan Layanan  : Selamat! Anda berhak mendapatkan Souvenir Gratis.")
elif validasi_1012:
    print("Catatan Layanan  : Terima kasih telah berkunjung.")
```

**Penjelasan :**

Program Sistem Loket Terpadu ini dirancang murni mematuhi instruksi penggunaan 4 pilar percabangan. Sesuai aturan, program sama sekali tidak menggunakan *nested-if* (if bersarang) maupun fungsi penghenti paksa, melainkan menggunakan logika linear yang dievaluasi secara runut. Berikut adalah rincian komponen kodenya:

1. **Penggunaan *Boolean Flag* (`validasi_1012`)**
   Program menggunakan variabel penanda (*flag*) yang diatur menjadi `True` pada awal eksekusi. Fungsi variabel ini adalah sebagai "saklar". Jika input pengguna tidak logis (seperti jumlah tiket minus atau nomor paket salah), maka saklar diubah menjadi `False`. Saklar ini kemudian disisipkan (di-*inject*) ke setiap blok percabangan di bawahnya menggunakan operator `and`, sehingga apabila nilainya `False`, sisa perhitungan di bawahnya akan otomatis dilewati tanpa memerlukan *nested-if*.

2. **Pilar 1: If Tunggal (Validasi Kelogisan Tiket)**
   Blok kode `if jumlah_tiket_1012 <= 0:` bertugas sebagai filter pertama. Jika pengguna memasukkan jumlah tiket 0 atau negatif, program hanya mencetak peringatan kuota tidak valid dan segera mengubah `validasi_1012 = False`.

3. **Pilar 2: Match-Case (Pemilihan Paket Wahana)**
   Fitur *Structural Pattern Matching* (`match-case`) digunakan untuk memetakan input angka `paket_1012` ke dalam detail harga dan nama wahana. 
   - `case 1` sampai `case 5`: Menginisialisasi variabel `nama_wahana_1012` dan `harga_satuan_1012` sesuai pilihan pengguna.
   - `case _`: Berfungsi sebagai nilai *default* (pengganti *else*). Jika pengguna menginput selain 1-5, program mencetak pesan tidak valid dan mengubah saklar validasi menjadi `False`.

4. **Pilar 3: If-Elif-Else Terstruktur (Validasi Izin Kendali Wahana)**
   Blok ini menyeleksi kelayakan usia dan kepemilikan SIM menggunakan operator logika majemuk (`and`, `!=`).
   - Khusus pada paket ke-3 (ATV Ekstrim), seluruh pengecekan usia (`umur_1012 >= 17`) dan SIM (`sim_1012 == 'y'`) diratakan dalam satu baris untuk menghindari *nested-if*.
   - Rangkaian menggunakan `elif` agar pemeriksaannya bersifat eksklusif (jika satu kondisi sudah benar, sisa kondisi di bawahnya tidak akan dieksekusi lagi).

5. **Pilar 4: Multi-If Terpisah (Akumulasi Diskon Bertingkat)**
   Bagian ini dengan sengaja dipisah menjadi rentetan `if` yang mandiri (tanpa `elif` atau `else`). Ini karena kita membutuhkan sifat komputasi yang akumulatif. Setiap kondisi (belanja >= 200.000, status *member*, ada kode promo, tiket >= 5) diperiksa satu per satu. Jika lebih dari satu kondisi yang terpenuhi, nilai `total_diskon_persen_1012` akan terus dijumlahkan (ditumpuk) hingga batas maksimal diskon.

6. **Pilar 5: If-Else Penutup (Audit & Pemformatan Output)**
   Setelah variabel diskon dan subtotal dihitung secara matematis, blok percabangan terakhir bertugas memverifikasi apakah total pembayaran melebihi batas Rp 300.000 untuk mendapatkan bonus *souvenir*. Selain itu, seluruh angka rupiah dicetak menggunakan metode *f-string* bawaan Python dengan parameter `{variabel:,.0f}`. Format ini secara otomatis menyisipkan koma sebagai pemisah ribuan dan membulatkan angka tanpa angka desimal di belakang koma, sehingga antarmuka terminal terlihat jauh lebih profesional.

# KESIMPULAN

Berdasarkan praktikum yang telah dilakukan, dapat disimpulkan bahwa penerapan struktur kendali percabangan adalah proses logis absolut di mana aplikasi tidak akan berjalan efisien tanpanya. Kita dapat memilah-milah validasi komputasi dengan `if`, `if-else`, dan `if-elif-else` demi alur yang dinamis. 

Dalam konteks pemrograman masa kini, Python telah mempermudah teknik perbandingan konstan melalui utilitas percabangan struktural `match-case` (diperkenalkan sejak Python 3.10) sehingga kode menjadi lebih rapi dan bersih. Keputusan antara penggunaan percabangan independen (untuk hasil akumulatif) dan percabangan terikat `elif` (untuk pilihan mutlak/eksklusif) memegang kunci dari arsitektur perangkat lunak yang fungsional dan terbebas dari malfungsi komputasi (*logical error*).
