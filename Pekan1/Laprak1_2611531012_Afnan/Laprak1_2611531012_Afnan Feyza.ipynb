# BAB II
# PEMBAHASAN

## 2.1 Dasar Teori
Dalam praktikum modul pertama ini, terdapat tiga komponen utama yang dipelajari:

1. **Python**: Bahasa pemrograman tingkat tinggi, *interpreted*, dan *general-purpose*. Python sangat menekankan pada keterbacaan kode (*code readability*).
2. **Visual Studio Code (VS Code)**: Editor teks kode sumber buatan Microsoft yang dapat dijalankan di berbagai sistem operasi. Editor ini mendukung berbagai bahasa pemrograman termasuk Python melalui instalasi ekstensi.
3. **GitHub**: Layanan berbasis *cloud* yang menyimpan sistem pengontrol versi (Git). Membantu *programmer* melacak perubahan kode dan berkolaborasi dalam tim.
4. **Fungsi Print dan Karakter Escape**: Fungsi dasar Python `print()` digunakan untuk mencetak *output* ke layar. Dalam memanipulasi *output* teks (*string*), dikenal karakter *escape* (menggunakan *backslash* `\`), seperti `\n` untuk baris baru atau `\\` untuk memunculkan simbol *backslash* itu sendiri.

## 2.2 Kode Program

### 2.2.1 Program "program1_2611531012"
Berikut adalah implementasi kode program pertama menggunakan sintaks dasar Python.

```python
# Program pertama
"""
Komentar
dua baris atau lebih
pertemuan 1 praktikum alpro
"""
print("Hello World")
print()
print("Program ini menghasilkan")
print("Empat baris output")
print("/ \\ // \\\\ /// \\\\\\")
print("Program ini" + " dibuat oleh a\n" + "Afnan Feyza_2611531012.")
```

**Penjelasan :**
1. **`""" ... """` (Docstring)**: Bagian pertama kode ini menggunakan tanda kutip tiga untuk membuat komentar multi-baris (*multiline comment*). Baris ini tidak akan dieksekusi oleh program dan berfungsi sebagai catatan.
2. **`print()`**: Fungsi dasar untuk mencetak teks ke layar. `print()` yang dibiarkan kosong akan mencetak baris kosong (sebagai jarak).
3. Karakter Escape `\\`**: Untuk mencetak simbol *backslash* (`\`) di Python, kita harus menggunakan *double backslash* (`\\`) agar tidak dibaca sebagai awalan perintah khusus.
4. **Konkatenasi `+` dan `\n`**: Pada baris terakhir, digunakan simbol `+` untuk menggabungkan (*concatenate*) beberapa *string*. Terdapat juga `\n` (newline) yang berfungsi untuk memaksa teks setelahnya turun ke baris baru, sehingga nama Afnan Feyza dicetak di baris bawah.
