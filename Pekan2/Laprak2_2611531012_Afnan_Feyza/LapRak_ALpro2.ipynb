## 2.1 Dasar Teori
1. **Tipe Data Numerik:** Python mendukung pengolahan data bilangan bulat (integer/`int`), desimal pecahan (`float`), maupun bilangan kompleks (`complex`) untuk kebutuhan algoritma kalkulasi aritmatika.
2. **Tipe Data String:** Digunakan untuk mendeklarasikan objek berbentuk teks. Dapat diapit kutip satu (`'`), dua (`"`), ataupun tiga (`"""`) apabila berisi banyak baris (multiline).
3. **Tipe Data Boolean:** Sebuah tipe data esensial yang hanya dapat menyimpan nilai `True` atau `False`, yang sering kali dimanfaatkan pada pembuatan cabang percabangan logika program.
4. **Input Data Dinamis:** Berbeda dari statis yang langsung dipatok ke variabel, perintah `input()` memungkinkan alur sistem meminta pengguna memberikan masukan teks.
5. **Konstanta:** Meski Python tidak mengekang deklarasi data, konstanta lazim digunakan untuk nilai yang disepakati mutlak dan dilarang berubah di tengah jalan, contohnya melalui bantuan dekorator `Final`.

## 2.2 Kode Program

### 2.2.1 Program "numerik_2611531012.py"
Program berikut mengilustrasikan penciptaan beberapa variabel, masing-masing dengan muatan tipe numerik yang berbeda (integer, float, complex), kemudian mengidentifikasi validasi struktur bawaannya memakai built-in function `type()`.

```python
# Buat file dengan nama numerik_2611531012.py
#crate a variable with integer value
a_1012=100
print("The type of variable having value", a_1012, " is ", type(a_1012))

#create a variable with float value.
b_1012=10.23455
print("The type of variable having value", b_1012, " is ", type(b_1012))

#create a variable with complex value.
c_1012=100+3j
print("The type of variable having value", c_1012, " is ", type(c_1012))
```

**Penjelasan :**
1. `a_1012 = 100` : Berfungsi untuk mendeklarasikan variabel `a_1012` dan menginisialisasinya dengan nilai bilangan bulat (integer), yaitu `100`.
2. `b_1012 = 10.23455` : Berfungsi untuk mendeklarasikan variabel `b_1012` dengan menugaskan nilai berupa pecahan atau desimal (float).
3. `c_1012 = 100+3j` : Berfungsi untuk mendeklarasikan variabel `c_1012` dengan nilai bilangan kompleks (complex) bawaan Python yang ditandai dengan sufiks `j`.
4. `type(...)` pada fungsi `print` : Berfungsi sebagai *built-in function* dari Python yang digunakan untuk mengecek tipe data aktual dari variabel yang ada di dalam tanda kurungnya, dan mencetak nama kelas tipe datanya ke layar terminal.

### 2.2.2 Program String "string_2611531012"
Blok program identitas diri di bawah ini mengandalkan string untuk mendefinisikan informasi profil. Teks dapat ditampung dalam sebaris tanda petik maupun kutipan multiline string ketika mendefinisikan lokasi alamat secara berderet.

```python
# Buat file dengan nama string_2611531012.py
#nama variabel ditambah 4 digit nim terakhir contoh: nama_1012
#ganti programnya dan data masing2
nama_1012 = "Afnan Feyza"
jenis_kelamin_1012 = 'L'
alamat_1012 = """
Jalan Kakak Tua No 2A
Kecamatan Padang Utara
Kota Padang
"""
agama_1012 = 'islam'
print("Nama saya adalah:", nama_1012)
print("Jenis kelamin saya adalah:", jenis_kelamin_1012)
print("Alamat saya adalah:", alamat_1012)
print("Agama saya adalah:", agama_1012)
```

Penjelasan program string di atas:

1. `nama_1012 = "Afnan Feyza"` dan `jenis_kelamin_1012 = 'L'` : Berfungsi menginisiasi variabel tipe data teks (string). Python mengizinkan pemakaian tanda kutip dua (`" "`) maupun tanda kutip tunggal (`' '`) untuk membungkus sebaris teks.
2. `alamat_1012 = """..."""` : Berfungsi untuk menampung teks yang berisi lebih dari satu baris atau *multiline string*. Teks dibungkus menggunakan tanda kutip tiga (bisa `"""` atau `'''`) agar perpindahan baris (enter) ikut tersimpan sebagai bagian dari data alamat tersebut.
3. `print("Nama saya adalah:", nama_1012)` : Berfungsi untuk mencetak gabungan string karakter statis dengan nilai yang tersimpan di dalam suatu variabel string.

### 2.2.3 Program Boolean "boolean_2611531012"
Program berikut mengaplikasikan boolean untuk mengecek syarat parameter nilai kelulusan. Status nilai akan dicari kepastiannya (`True` atau `False`) sebelum divalidasi dan diizinkan melaju memanggil perintah cetak Cum Laude.

```python
# Buat file dengan nama boolean_2611531012.py
# Nama variabel ditambah 4 digit nim terakhir contoh: nama_1012
# Deklarasi variabel dengan tipe data boolean

is_lulus_1012 = True
is_cumlaude_1012 = True

# Menggunakan Boolean
nilai_1012 = 85
batas_lulus_1012 = 75

# Menentukan nilai boolean dari kondisi
status_kelulusan_1012 = nilai_1012 >= batas_lulus_1012 # Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai:", nilai_1012)
print("Apakah Lulus?:" , status_kelulusan_1012)
if is_lulus_1012 and is_cumlaude_1012:
    print("Selamat! Anda lulus dengan predikat Cum Laude!")
```

Penjelasan program boolean di atas:

1. `is_lulus_1012 = True` dan `is_cumlaude_1012 = True` : Berfungsi mendeklarasikan variabel tipe data Boolean yang merepresentasikan kebenaran, di mana nilainya hanya dapat diatur sebagai `True` (benar) atau `False` (salah).
2. `status_kelulusan_1012 = nilai_1012 >= batas_lulus_1012` : Berfungsi melakukan evaluasi operator perbandingan relasional (lebih besar sama dengan). Karena nilai 85 lebih besar dari 75, hasil evaluasi tersebut akan mengembalikan nilai kebenaran Boolean berupa `True` lalu disalurkan ke variabel `status_kelulusan_1012`.
3. `if is_lulus_1012 and is_cumlaude_1012:` : Berfungsi sebagai blok pengkondisian menggunakan operator logika `and`. Program hanya akan mengeksekusi fungsi print berupa ucapan "Selamat!" yang ada di bawahnya apabila kondisi di kiri dan kanan sintaks `and` sama-sama bernilai `True`.

### 2.2.4 Program Input "input_2611531012"
Skrip program ini bergantung pada `input()` agar sisi pengerjaan perumusan dapat secara fleksibel disuplai masukan dari sisi end-user. Masukan awal direkam menggunakan string default untuk lalu dipaksa berubah bentuk (`type casting`) menjadi bilangan `int` agar kalkulasi dimensi berjalan lancar.

```python
# Buat file dengan nama input_26115531012.py
# Buat program untuk menghitung luas persegi panjang
# Nama variabel ditambah 4 digit nim terakhir contoh: panjang_1012
# Program ini menggunakan fungsi input() untuk meminta pengguna
# memasukkan nilai panjang dan lebar persegi panjang.
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
panjang_1012 = input('Masukkan nilai panjang: ')
lebar_1012 = input('Masukkan nilai lebar: ')
luas_1012 = int(panjang_1012) * int(lebar_1012)
print("Luas:", luas_1012)
```

**Penjelasan :**
1. `input('Masukkan nilai panjang: ')` : Berfungsi menampilkan teks sapaan ke layar, kemudian menahan eksekusi program sejenak untuk menunggu pengguna mengetikkan sesuatu (inputan) ke dalam *console* / *terminal*. Secara bawaan (default), semua masukan akan direkam sebagai data String.
2. `int(panjang_1012)` dan `int(lebar_1012)` : Berfungsi melakukan proses *type casting* atau pemaksaan konversi tipe data.[cite: 2] Nilai panjang dan lebar yang awalnya berupa karakter teks atau String diubah paksa menjadi bilangan bulat (Integer) agar dapat dikalikan dalam operasi matematika `*` (perkalian).
3. `luas_1012 = ...` : Menyimpan hasil akhir nilai perkalian yang bertipe Integer.

### 2.2.5 Program Konstanta "konstanta_2611531012"
Menggunakan import atribut `Final` dari modul `typing`, kode ini mempatenkan variabel global `PI` senilai 3.14. Tujuannya adalah mereduksi kesalahan perubahan nilai sembarang secara tidak sengaja ketika program dikomputasi untuk menelusuri luas lingkaran.

```python
# Buat file dengan nama konstanta_2611531012.py
# Program ini menggunakan konstanta untuk menghitung luas lingkaran.
# Nama variabel ditambah 4 digit nim terakhir contoh: jari_1012

from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_1012 = float(input("Masukkan nilai jari-jari: "))
luas_1012 = PI * jari_1012 * jari_1012
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_1012, luas_1012))
```

**Penjelasan :**
1. `from typing import Final` : Berfungsi untuk memanggil library atau modul bawaan `typing` dari Python, guna menggunakan kemampuan pelabelan (anotasi) jenis variabel tertentu dalam hal ini sebagai tipe `Final`.
2. `PI: Final = 3.14` : Mendeklarasikan bahwa variabel `PI` bersifat final atau konstan. Ini menunjukkan pada struktur kode bahwa `PI` merupakan tetapan nilai matematis yang tidak boleh diubah oleh baris kode mana pun secara tidak sengaja di masa mendatang.
3. `float(input(...))` : Meminta input dari pengguna, namun berkat pembungkusan fungsi tipe data `float(...)`, teks angka yang dimasukkan pengguna secara sekejap langsung dikonversi ke tipe bilangan desimal (float).
4. `%f` dan `%.2f` : Berfungsi sebagai teknik pemformatan string kuno gaya bahasa C (*String Formatting*). Simbol `%f` mencetak bilangan float secara utuh, sedangkan `%.2f` digunakan untuk membatasi tampilan hasil hitung luas dan input jari-jari agar hanya menyisakan dua digit angka spesifik saja setelah tanda koma.

## Tugas Praktikum

**Soal :**
"**Sistem Biodata dan Validasi Kelulusan Praktikan**",
Rancang sebuah program konsol sederhana yang mencatat data identitas praktikan baru dan melakukan verifikasi status kelulusan awal praktikum secara otomatis dengan syarat:

1. Nama berkas `Tugas2_2611531012.py` dan seluruh variabel diakhiri dengan 4 digit terakhir NIM (`_1012`).

2. Terapkan konstanta batas kelulusan menggunakan modul `typing.Final`.

3. Terapkan tipe data String & Char (untuk nama, jenis kelamin, dan multiline string untuk alamat).

4. Terapkan tipe data Numerik & Type Casting (`int` untuk umur, `float` untuk skor, dan `complex` untuk token identifikasi).

5. Evaluasi kelulusan secara dinamis menjadi tipe data Boolean dan tampilkan seluruh tipe data ke layar menggunakan fungsi bawaan `type()`

**Kode Jawaban :**

a. Kode Program "Tugas2_2611531012.py"

```python
# Kode program tugas "Tugas2_2611531012.py"

from typing import Final

# Konstanta batas kelulusan
BATAS_LULUS: Final = 75.0

print("=== SISTEM REGISTRASI PRAKTIKAN ALPRO 2026 ===")
nama_1012 = input("Masukkan Nama Mahasiswa\t\t: ")
jenis_kelamin_1012 = input("Masukkan Jenis Kelamin (L/P)\t: ")
umur_1012 = int(input("Masukkan Umur\t\t\t: "))
skor_tes_1012 = float(input("Masukkan Skor Tes Awal\t\t: "))

# Deklarasi alamat secara multiline
alamat_1012 = """Jl. Kakak Tua No 2, 
Kecamatan Padang Utara, Kelurahan Air Tawar Barat,
Kota Padang"""

# Token identifikasi menggunakan bilangan kompleks
token_1012 = 100+3j

# Evaluasi status kelulusan
status_lulus_1012 = skor_tes_1012 >= BATAS_LULUS

print("\n=== DATA PRAKTIKAN & HASIL PEMERIKSAAN ===")
print("Nama Mahasiswa\t:", nama_1012, "| Tipe:", type(nama_1012))
print("Jenis Kelamin\t:", jenis_kelamin_1012, "| Tipe:", type(jenis_kelamin_1012))
print("Alamat Domisili\t:\n" + alamat_1012, "\n| Tipe:", type(alamat_1012))
print("Umur\t\t:", umur_1012, "tahun | Tipe:", type(umur_1012))
print("Skor Tes Awal\t:", skor_tes_1012, "| Tipe:", type(skor_tes_1012))
print("ID Token Sinyal\t:", token_1012, "| Tipe:", type(token_1012))

print("\n=== STATUS KELULUSAN PRAKTIKUM ===")
print("Batas Minimum Nilai:", BATAS_LULUS)
print("Apakah Dinyatakan Lulus?:", status_lulus_1012, "| Tipe:", type(status_lulus_1012))
```
